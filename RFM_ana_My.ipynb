```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("RFM.csv")
df.head()
```

```python
# 1. 確保 'Date' 欄位是 datetime 型態，以便後續做Date計算
df['Date'] = pd.to_datetime(df['Date'])

# 2. 設定 R（Recency）計算的基準日（通常可設定為資料集裡的最大Date，或是當下Date pd.Timestamp('today')）
# 這裡採用資料集中的最大Date作為基準點，你也可以根據需求自行替換
max_date = df['Date'].max()

# 3. 進行 groupby 並計算 RFM
result_df = (
    df.groupby('CustomerKey').agg(
    F=('InvoNo', 'count'),  # InvoNo計數
    M=('Amount', 'sum'),  # Amount加總
    # 算每一筆Date與該客戶最大交易Date的差額（單位：天），若只要找整個資料集最大Date的差額請看下方註解
    max_date=('Date', 'max'),)
.reset_index())


# 4. 計算 R：全體最大Date（或基準日）與該客戶最大交易日的天數差
result_df['R'] = (max_date - result_df['max_date']).dt.days

result_df=result_df.rename(columns={'CustomerKey':'Customer'})[['Customer','F','M','R']]

# 檢視結果
print(result_df)
```
